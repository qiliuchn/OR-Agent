## Coordinated Idea Generation

In contrast, OR-Agent employs coordinated generation, where all child ideas of a node are produced in a single LLM invocation. The model is explicitly instructed to generate multiple \emph{distinct} directions simultaneously. Using the same BPP-Online parent solution, twenty coordinated ideas are shown below.

1. Normalize the prize-to-weight ratio per constraint dimension and aggregate via geometric mean to better reflect balanced resource usage across all constraints.
2. Introduce a sparsification step that sets heuristic values to zero for items whose total weight exceeds a threshold (e.g., average capacity usage), focusing selection on feasible candidates.
3. Incorporate a penalty term based on the maximum weight of an item across all constraints to discourage selection of items that dominate any single resource.
4. Compute heuristic as prize divided by the L2 norm of the weight vector, emphasizing compactness in multi-dimensional weight space.
5. Use min-max normalization of weights per dimension before computing prize-to-weight ratios to ensure equitable contribution from each constraint.
6. Define heuristic as the product of prize and the inverse of the maximum weight across constraints, prioritizing high-value, low-peak-load items.
7. Apply a logarithmic transformation to prize and/or weight to reduce the impact of extreme values and improve scaling behavior.
8. Combine multiple heuristics (e.g., sum of ratios, max ratio, L2-based ratio) into a weighted ensemble, where weights are tuned or fixed based on problem structure.
9. Introduce a feasibility-aware heuristic: set heuristic to zero if any individual weight exceeds 1.0 (the capacity limit per dimension), preemptively excluding infeasible items.
10. Compute a 'slack-aware' heuristic by estimating remaining capacity margin and favoring items that fit well within typical slack patterns observed in elite solutions.
11. Use a normalized prize-to-weight ratio where weights are scaled by the inverse of the number of constraints, adjusting for dimensionality bias.
12. Implement a two-stage heuristic: first rank items by prize density, then re-rank by compatibility with other high-density items using pairwise weight correlation.
13. Derive heuristic from a surrogate linear programming relaxation score, approximating marginal utility of including each item.

16. Weight the contribution of each constraint dimension inversely by its global tightness (e.g., average weight per item in that dimension), giving more importance to scarcer resources.
17. Replace sum of weights with harmonic mean of weights to penalize items with highly uneven weight distributions across constraints.
18. Use a robust statistic like median instead of sum or mean when aggregating weights to reduce sensitivity to outliers in weight dimensions.

### Advantages and Trade-offs

**Advantages of Coordinated Generation**
- **Improved diversity**: Ideas exhibit greater conceptual separation.
- **Better coverage**: Exploration spans multiple heuristic paradigms.
- **Systematic reasoning**: The model implicitly allocates different ``slots'' to distinct improvement themes.

**Limitations**
Coordinated generation is not without drawbacks:
- Individual ideas tend to be shorter when many are generated simultaneously.
- Explicitly constraining word count per idea is possible but inelegant, as optimal length is task-dependent.
- Generating many ideas in one call is constrained by the context window.

Thus, coordinated generation introduces a trade-off between diversity and per-idea elaboration.



### Implications for Research Tree Expansion

Within OR-Agent, each research tree node expands via coordinated generation. All child hypotheses are produced together, evaluated separately, and selectively retained based on performance. This design reduces redundant branching and improves structural exploration efficiency. While independent generation can be seen as Monte Carlo sampling in local neighborhoods, coordinated generation resembles structured hypothesis decomposition. Empirical results across multiple problems indicate that coordinated generation improves exploration breadth without significantly increasing computational overhead, thereby contributing to more stable research progression.
